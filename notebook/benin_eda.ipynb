## outlier detection using IQR

IQR
==>Considers values outside Q1 - 1.5IQR and Q3 + 1.5IQR as outliers
==>Works well for skewed distributions; robust to extreme values
==>Best choice — robust for irradiance data and non-normal distribution

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert Timestamp to datetime
df['Timestamp'] = pd.to_datetime(df['Timestamp'])

# Irradiance columns
irradiance_cols = ['GHI', 'DNI', 'DHI']

# Step 1: Replace negatives with zero
df[irradiance_cols] = df[irradiance_cols].clip(lower=0)

# Step 2: Exclude zeros
daylight_df = df[(df['GHI'] > 0) & (df['DNI'] > 0) & (df['DHI'] > 0)]

# Step 3: Detect outliers using IQR
outliers_dict = {}
for col in irradiance_cols:
    Q1 = daylight_df[col].quantile(0.25)
    Q3 = daylight_df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = daylight_df[(daylight_df[col] < Q1 - 1.5*IQR) | (daylight_df[col] > Q3 + 1.5*IQR)]
    outliers_dict[col] = outliers
    print(f"{col}: {len(outliers)} outliers detected")

# Step 4: Plot histograms excluding zeros
sns.set_style('darkgrid')
plt.figure(figsize=(15, 5))

for i, col in enumerate(irradiance_cols, 1):
    plt.subplot(1, 3, i)
    sns.histplot(daylight_df[col], bins=50, kde=True, color='steelblue', edgecolor='black')
    plt.title(f'{col} Distribution (Daylight, Outliers Included)')
    plt.xlabel(f'{col} (W/m²)')
    plt.ylabel('Frequency')

plt.tight_layout()
plt.show()


```

## outliers excluded 


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert Timestamp to datetime if not already
df['Timestamp'] = pd.to_datetime(df['Timestamp'])

# Irradiance columns
irradiance_cols = ['GHI', 'DNI', 'DHI']

# Step 1: Replace negatives with zero
df[irradiance_cols] = df[irradiance_cols].clip(lower=0)

# Step 2: Exclude zeros (nighttime)
daylight_df = df[(df['GHI'] > 0) & (df['DNI'] > 0) & (df['DHI'] > 0)].copy()

# Step 3: Remove outliers using IQR
for col in irradiance_cols:
    Q1 = daylight_df[col].quantile(0.25)
    Q3 = daylight_df[col].quantile(0.75)
    IQR = Q3 - Q1
    # Keep only values within the IQR range
    daylight_df = daylight_df[
        (daylight_df[col] >= Q1 - 1.5 * IQR) &
        (daylight_df[col] <= Q3 + 1.5 * IQR)
    ]

print(f"Cleaned dataset shape: {daylight_df.shape}")

# ✅ Step 4: Compute and print averages (after cleaning and removing outliers)
mean_values = daylight_df[irradiance_cols].mean()
print("\n=== Average Irradiance Values After Cleaning ===")
for col, val in mean_values.items():
    print(f"{col}: {val:.2f} W/m²")

# Step 5: Plot histograms of cleaned irradiance data
sns.set_style('darkgrid')
plt.figure(figsize=(15, 5))

for i, col in enumerate(irradiance_cols, 1):
    plt.subplot(1, 3, i)
    sns.histplot(daylight_df[col], bins=50, kde=True, color='steelblue', edgecolor='black')
    plt.title(f'{col} Distribution (Cleaned Daylight)')
    plt.xlabel(f'{col} (W/m²)')
    plt.ylabel('Frequency')

plt.tight_layout()
plt.show()

```

## Evaluating Solar Potential

```python
# --- Solar Data Temporal Analysis Pipeline ---
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

# =====================
# 1️⃣ Load and Prepare Data
# =====================
df = pd.read_csv("../data/benin-malanville.csv")

# Ensure Timestamp column exists and is in datetime format
if 'Timestamp' not in df.columns:
    df['Timestamp'] = pd.date_range(start='2024-01-01', periods=len(df), freq='T')

df['Timestamp'] = pd.to_datetime(df['Timestamp'])
df = df.set_index('Timestamp')

# =====================
# 2️⃣ Resampling to Daily and Monthly Averages
# =====================
daily = df.resample('D').mean()
monthly = df.resample('M').mean()

# =====================
# 3️⃣ Visualize Daily GHI Trend
# =====================
plt.figure(figsize=(10, 5))
plt.plot(daily['GHI'], color='orange', label='Daily GHI')
plt.title('Daily Average Global Horizontal Irradiance (GHI)')
plt.xlabel('Date')
plt.ylabel('GHI (W/m²)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# =====================
# 4️⃣ Monthly GHI and Temperature Trends
# =====================
plt.figure(figsize=(10, 5))
plt.plot(monthly['GHI'], label='GHI (Irradiance)', color='gold')
plt.plot(monthly['Tamb'], label='Ambient Temp (°C)', color='tomato')
plt.plot(monthly['TModA'], label='Module A Temp (°C)', color='royalblue')
plt.plot(monthly['TModB'], label='Module B Temp (°C)', color='green')
plt.title('Monthly Trends: Irradiance and Temperatures')
plt.xlabel('Month')
plt.ylabel('Value')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# =====================
# 5️⃣ Correlation Analysis
# =====================
corr = df[['GHI', 'DNI', 'DHI', 'Tamb', 'TModA', 'TModB', 'RH', 'WS', 'Precipitation']].corr()
print("\n🔍 Correlation Matrix:")
print(corr['GHI'].sort_values(ascending=False))

# =====================
# 6️⃣ Solar Suitability Scoring (Temporal)
# =====================
# Prepare data for scoring
features = ['GHI', 'Tamb', 'RH', 'WS', 'Precipitation']
data = monthly[features].copy()

# Normalize using MinMaxScaler
scaler = MinMaxScaler()
scaled = pd.DataFrame(scaler.fit_transform(data), columns=data.columns, index=data.index)

# Weighted solar suitability score
scaled['SolarScore'] = (
    0.4 * scaled['GHI'] +
    0.2 * (1 - scaled['Tamb']) +
    0.2 * (1 - scaled['RH']) +
    0.1 * (1 - scaled['Precipitation']) +
    0.1 * scaled['WS']
)

# Rank the months
ranked = scaled.sort_values(by='SolarScore', ascending=False)
print("\n🌞 Top Months for Solar Potential:")
print(ranked[['SolarScore']].head())

# =====================
# 7️⃣ Plot Solar Score Over Time
# =====================
plt.figure(figsize=(10, 5))
plt.plot(scaled.index, scaled['SolarScore'], marker='o', color='darkorange')
plt.title('Monthly Solar Suitability Score')
plt.xlabel('Month')
plt.ylabel('Score (0–1)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# =====================
# 8️⃣ Summary for Strategy
# =====================
print("\n✅ STRATEGIC INSIGHT:")
print("- High GHI and moderate temperatures indicate optimal solar conditions.")
print("- Months with top SolarScore are most suitable for targeted solar investments.")
print("- Low humidity and precipitation improve efficiency and system reliability.")
print("- Use these insights to plan panel maintenance and optimize energy storage.")
```

## seeing if there is a relationship between the variables and the cleaness of the reading modules 

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind

# Read your data (replace with your actual path)
df = pd.read_csv("../data/benin-malanville.csv")

# Make sure Cleaning is integer or boolean
df['Cleaning'] = df['Cleaning'].astype(int)

# Define the irradiance columns
irradiance_cols = ['GHI', 'DNI', 'DHI', 'ModA', 'ModB']

# Separate data based on cleaning status
cleaned = df[df['Cleaning'] == 1]
not_cleaned = df[df['Cleaning'] == 0]

# Compare average values
print("=== Average Irradiance Values ===")
comparison = pd.DataFrame({
    "Not Cleaned": not_cleaned[irradiance_cols].mean(),
    "Cleaned": cleaned[irradiance_cols].mean()
})
comparison["Difference"] = comparison["Cleaned"] - comparison["Not Cleaned"]
print(comparison.round(2))

# Visualize the differences
comparison[["Not Cleaned", "Cleaned"]].plot(kind="bar", figsize=(8,5))
plt.title("Average Irradiance Before vs After Cleaning")
plt.ylabel("Irradiance (W/m²)")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()

# Optional: statistical test
print("\n=== T-test Results (Are they significantly different?) ===")
for col in irradiance_cols:
    t_stat, p_val = ttest_ind(
        cleaned[col].dropna(),
        not_cleaned[col].dropna(),
        equal_var=False
    )
    print(f"{col}: p-value = {p_val:.4f}")
```

## TmodA and TmodB when the sensor module is clean and when not clean

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load your dataset (update file path if needed)
df = pd.read_csv("../data/benin-malanville.csv")

# Ensure Cleaning is numeric
df['Cleaning'] = df['Cleaning'].astype(int)

# Calculate mean TModA and TModB for each cleaning status
temp_means = df.groupby('Cleaning')[['TModA', 'TModB']].mean().round(2)

# Bar chart setup
x = np.arange(len(temp_means.columns))  # positions for TModA, TModB
width = 0.35

fig, ax = plt.subplots(figsize=(8,5))

# Plot bars for Cleaning = 0 and Cleaning = 1
ax.bar(x - width/2, temp_means.loc[0], width, label='Not Cleaned', color='steelblue')
ax.bar(x + width/2, temp_means.loc[1], width, label='Cleaned', color='orange')

# Chart formatting
ax.set_title('Average Module Temperatures Before vs After Cleaning')
ax.set_ylabel('Temperature (°C)')
ax.set_xticks(x)
ax.set_xticklabels(temp_means.columns)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7)

# Add value labels on bars
for i, v in enumerate(temp_means.loc[0]):
    ax.text(i - width/2, v + 0.5, f'{v:.1f}', ha='center', fontsize=9)
for i, v in enumerate(temp_means.loc[1]):
    ax.text(i + width/2, v + 0.5, f'{v:.1f}', ha='center', fontsize=9)

plt.tight_layout()
plt.show()
```

## cleaning the data

```python
import pandas as pd
import os

# === 1. Load the dataset ===
input_path = "../data/benin-malanville.csv"   # Source file
output_path = "../cleaned/benin-cleaned.csv"  # Destination file

# Read the CSV file
df = pd.read_csv(input_path)

# === 2. Drop the 'Comments' column if it exists ===
if 'Comments' in df.columns:
    df = df.drop(columns=['Comments'])

# === 3. Keep only rows where Cleaning == 1 ===
if 'Cleaning' in df.columns:
    df = df[df['Cleaning'] == 1]
else:
    raise KeyError("The dataset doesn't have a 'Cleaning' column.")

# === 4. Keep only rows where irradiance values are non-negative ===
irradiance_vars = ['GHI', 'DNI', 'DHI', 'ModA', 'ModB']

# Filter condition (only rows with all irradiance values ≥ 0)
df = df[(df[irradiance_vars] >= 0).all(axis=1)]

# === 5. Make sure the output directory exists ===
os.makedirs(os.path.dirname(output_path), exist_ok=True)

# === 6. Save the cleaned data ===
df.to_csv(output_path, index=False)

print(f"✅ Cleaned dataset saved to: {output_path}")
print(f"Remaining rows: {len(df)}")
```

## imputation of "cleaning" variable

```python
import pandas as pd
import numpy as np

# Load your data
df = pd.read_csv("../data/benin-malanville.csv")

# Convert Timestamp to datetime
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
df['Hour'] = df['Timestamp'].dt.hour

irradiance_cols = ['GHI', 'DNI', 'DHI']

# Step 1: Replace negatives with zero
df[irradiance_cols] = df[irradiance_cols].clip(lower=0)

# Step 2: Mark "daytime" rows (roughly 6 AM to 6 PM)
df['is_daylight'] = df['Hour'].between(6, 18)

# Step 3: Define a rule for "realistic" irradiance
# Example: GHI > 50 and DNI > 50 (tunable threshold)
df['realistic_irradiance'] = (
    (df['GHI'] > 50) & 
    (df['DNI'] > 50) & 
    (df['DHI'] >= 0)
)

# Step 4: Keep rows that are either:
# - Cleaned (Cleaning == 1)
# - OR uncleaned but realistic and during daylight
filtered_df = df[(df['Cleaning'] == 1) | 
                 ((df['Cleaning'] == 0) & df['is_daylight'] & df['realistic_irradiance'])]

# Step 5: Add a confidence weight column
# Higher weight if cleaned, lower if not
filtered_df['confidence'] = np.where(filtered_df['Cleaning'] == 1, 1.0, 0.5)

print(f"Original data size: {len(df)}")
print(f"Filtered data size: {len(filtered_df)}")
print(f"Percentage retained: {len(filtered_df) / len(df) * 100:.2f}%")

# Optional: save the filtered data
filtered_df.to_csv("../cleaned/benin-filtered.csv", index=False)
print("Filtered dataset saved.")
```

## difference between what was lost during cleaning and what got recovered after smart filtering.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load your data
df = pd.read_csv("../data/benin-malanville.csv")

# Convert Timestamp to datetime
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
df['Hour'] = df['Timestamp'].dt.hour

irradiance_cols = ['GHI', 'DNI', 'DHI']

# Step 1: Replace negatives with zero
df[irradiance_cols] = df[irradiance_cols].clip(lower=0)

# Step 2: Mark "daytime" rows (approx 6 AM to 6 PM)
df['is_daylight'] = df['Hour'].between(6, 18)

# Step 3: Define a rule for "realistic" irradiance
df['realistic_irradiance'] = (
    (df['GHI'] > 50) &
    (df['DNI'] > 50) &
    (df['DHI'] >= 0)
)

# Step 4: Keep rows that are either:
# Cleaned (cleaning == 1)
# OR uncleaned but realistic and during daylight
filtered_df = df[
    (df['Cleaning'] == 1) |
    ((df['Cleaning'] == 0) & df['is_daylight'] & df['realistic_irradiance'])
]

# Step 5: Add a confidence weight column
filtered_df['confidence'] = np.where(filtered_df['Cleaning'] == 1, 1.0, 0.5)

# --- Summary statistics ---
print(f"Original data size: {len(df)}")
print(f"Filtered data size: {len(filtered_df)}")
print(f"Percentage retained: {len(filtered_df) / len(df) * 100:.2f}%")

# --- Visualization section ---
sns.set_style("whitegrid")
plt.figure(figsize=(15, 5))

for i, col in enumerate(irradiance_cols, 1):
    plt.subplot(1, 3, i)
    sns.kdeplot(df[col], label='Original', color='red', linewidth=1.5)
    sns.kdeplot(filtered_df[col], label='Filtered', color='green', linewidth=1.5)
    plt.title(f'{col} Distribution: Before vs After Filtering')
    plt.xlabel(f'{col} (W/m²)')
    plt.legend()

plt.tight_layout()
plt.show()

# --- Save filtered data ---
filtered_df.to_csv("../filtered/benin-filtered.csv", index=False)
print("Filtered dataset saved to '../cleaned/benin-filtered.csv'")
```
